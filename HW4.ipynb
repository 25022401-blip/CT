Viết hàm truyền vào hai số nguyên a, b. Trả về số lớn hơn trong hai số. def Max_Of_Two( a, b) ->
int: 


```python
def Max_Of_Two(a,b):
    return max(a,b)
x,y=map(int,input().split())
Max_Of_Two(x,y)
```

Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ
bằng tổng hai số trước đó
Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt
quá A 

```python
A=int(input())
a,b=1,1
while b<=A:
    a,b=b,a+b
print(a)    
```
